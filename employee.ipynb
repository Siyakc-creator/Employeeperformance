Real-World Problem: Analyze Employee Performance Using NumPy
Problem Statement
A company wants to analyze the monthly performance of its employees. The HR department has collected the performance scores of 6 employees for 4 months.
Each employee is evaluated on a scale of 0–100.
Using Python and NumPy, create a program that:
Stores the employee performance data in a NumPy array.
Calculates the average performance score of each employee.
Finds the highest and lowest monthly score for each employee.
Calculates the overall average score of all employees.
Identifies employees whose average score is 80 or above.
Finds which employee has the highest overall performance.
Assigns a performance category:
90+ → Excellent
80–89 → Very Good
70–79 → Good
Below 70 → Needs Improvement

| Employee | Month 1 | Month 2 | Month 3 | Month 4 |
| -------- | ------: | ------: | ------: | ------: |
| Ram      |      85 |      78 |      92 |      88 |
| Sita     |      72 |      80 |      75 |      79 |
| Hari     |      95 |      91 |      89 |      94 |
| Gita     |      65 |      70 |      68 |      72 |
| Krishna  |      88 |      84 |      90 |      86 |
| Mina     |      76 |      82 |      79 |      85 |

```python
import numpy as np 
```

```python
employees = np.array([
    "Ram", "Sita", "Hari", "Gita", "Krishna", "Mina"
])
print(employees)
```

```python
scores = np.array([
    [85, 78, 92, 88],
    [72, 80, 75, 79],
    [95, 91, 89, 94],
    [65, 70, 68, 72],
    [88, 84, 90, 86],
    [76, 82, 79, 85]
])
print(scores)
```

```python
average_scores = np.mean(scores, axis=1)
print("Average Scores",average_scores)
```

```python
highest_scores = np.max(scores, axis=1)
print("Higest Scores ",highest_scores)
```

```python
lowest_scores = np.min(scores, axis=1)
print("Lowest Scores ",lowest_scores)
```

```python
overall_average = np.mean(scores)
print("overall Average",round(overall_average,2))
```

```python
high_performers = employees[average_scores >= 80]
```

```python
best_index = np.argmax(average_scores)
print("Best Index ",best_index)
```

```python
best_employee = employees[best_index]
print("Best employee",best_employee)
```

```python
best_score = average_scores[best_index]
print("best Score",best_score)
```

```python
print("Employee Performance Report")
print("-" * 40)
```

```python
for i in range(len(employees)):
    if average_scores[i] >= 90:
        category = "Excellent"
    elif average_scores[i] >= 80:
        category = "Very Good"
    elif average_scores[i] >= 70:
        category = "Good"
    else:
        category = "Needs Improvement"

    print(
        f"{employees[i]}: "
        f"Average={average_scores[i]:.2f}, "
        f"Highest={highest_scores[i]}, "
        f"Lowest={lowest_scores[i]}, "
        f"Category={category}"
    )
```

```python
print("\nOverall Average:", round(overall_average, 2))
```

```python
print("\nHigh Performers:")
print(high_performers)
```

```python
print("\nBest Performer:")
print(best_employee, "-", round(best_score, 2))
```
